```python
import pandas as pd
import numpy as np

# Sample data
sample = pd.DataFrame({
    'Department': ['Sales', 'Sales', 'IT', 'IT', 'HR'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Salary': [50000, 60000, 70000, 80000, 55000]
})

# # Calculate the average salary of each department and match it to each employee
sample['Dept_Avg_Salary'] = sample.groupby('Department').Salary.transform('mean')

print(sample)
```
