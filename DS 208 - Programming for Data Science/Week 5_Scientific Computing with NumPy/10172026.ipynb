```python
import numpy as np

rng = np.random.default_rng(1)
scores = rng.normal(75, 10, size=100_000)   # made-up exam scores
data = scores.tolist()                       # the same numbers as a list

def by_hand():
    mean = sum(data) / len(data)
    var = sum((x - mean) ** 2 for x in data) / len(data)
    return mean, var ** 0.5

def with_numpy():
    return scores.mean(), scores.std()

m1, s1 = by_hand()
m2, s2 = with_numpy()
print(f"by hand: mean {m1:.4f}, std {s1:.4f} --(by_hand)")
print(f"numpy  : mean {m2:.4f}, std {s2:.4f} --(with_numpy)")
```

```python
import numpy as np

rng = np.random.default_rng(5)
m = rng.integers(0, 100, size=(4, 3))   # 4 rows, 3 columns

col_means = m.mean(axis=0)              # shape (3,)
centred = m - col_means                 # (4, 3) minus (3,)

print(m.shape, col_means.shape, centred.shape)
print(centred)
print(centred.sum(axis=0))              # each column now sums to 0
```

```python
rng = np.random.default_rng(101)

scores = rng.normal(75, 12, 50)

print(scores.mean())
print(scores.std())
print(scores.max())
print(scores.min())
print(scores.argmax())
print(scores.argmin())

con = np.where(scores >= 75, "Pass", "Fail")
absolute = np.abs(scores)
students_score_position = np.argsort(absolute)
students_score = np.sort(absolute)
new_scores = students_score.reshape((10,5))
print(new_scores)

column_sum = new_scores.sum(axis=0)
row_sum = new_scores.sum(axis=1)

print(f"Sums per column: {column_sum}")
print(f"Sums per row: {row_sum}")
```
